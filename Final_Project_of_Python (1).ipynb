## Stage 2 – Data Cleaning and Preparation

- Handled missing values in both datasets.
- Standardized column names between Amazon and Flipkart datasets.
- Removed unnecessary columns.
- Filled missing values using appropriate methods.
- Converted data types for analysis.
- Created a clean combined dataset.


```python
# Convert columns to numeric
Amazon["Quantity"] = pd.to_numeric(
    Amazon["Quantity"],
    errors="coerce"
)

Amazon["Unit_Price_INR"] = pd.to_numeric(
    Amazon["Unit_Price_INR"],
    errors="coerce"
)

Amazon["Total_Sales_INR"] = pd.to_numeric(
    Amazon["Total_Sales_INR"],
    errors="coerce"
)

Amazon["Review_Rating"] = pd.to_numeric(
    Amazon["Review_Rating"],
    errors="coerce"
)

# Fill Unit Price by product average
Amazon["Unit_Price_INR"] = Amazon.groupby(
    "Product_Name"
)["Unit_Price_INR"].transform(
    lambda x: x.fillna(x.mean())
)


# Fill remaining Unit Price with overall average
Amazon["Unit_Price_INR"] = Amazon["Unit_Price_INR"].fillna(
    Amazon["Unit_Price_INR"].mean()
)


# Fill Total Sales by product average
Amazon["Total_Sales_INR"] = Amazon.groupby(
    "Product_Name"
)["Total_Sales_INR"].transform(
    lambda x: x.fillna(x.mean())
)


# Fill remaining Total Sales with overall average
Amazon["Total_Sales_INR"] = Amazon["Total_Sales_INR"].fillna(
    Amazon["Total_Sales_INR"].mean()
)

Amazon["Review_Rating"] = Amazon["Review_Rating"].fillna(
    Amazon.groupby("Product_Name")["Review_Rating"].transform("mean")
)
# Check null values
Amazon.isnull().sum()
```

```python
Amazon["Product_Name"] = Amazon["Product_Name"].fillna("Unknown")

Amazon["City"] = Amazon["City"].fillna("Unknown")

Amazon["Payment_Method"] = Amazon["Payment_Method"].fillna("Unknown")

# Create product-category mapping from existing values

product_category_map = (
    Amazon.dropna(subset=["Product_Category"])
    .groupby("Product_Name")["Product_Category"]
    .first()
)

# Fill missing category using product name

Amazon["Product_Category"] = Amazon.apply(
    lambda row: product_category_map[row["Product_Name"]]
    if pd.isna(row["Product_Category"])
    and row["Product_Name"] in product_category_map
    else row["Product_Category"],
    axis=1
)

Amazon.isnull().sum()
```

```python
# ===============================
# FLIPKART DATA CLEANING
# ===============================


# 1. Check initial null values

flipkart.isnull().sum()



# ===============================
# 2. Fix data types
# ===============================

import pandas as pd
import numpy as np


flipkart["Quantity"] = pd.to_numeric(
    flipkart["Quantity"],
    errors="coerce"
)

flipkart["Unit_price"] = pd.to_numeric(
    flipkart["Unit_price"],
    errors="coerce"
)

flipkart["Total_Sales"] = pd.to_numeric(
    flipkart["Total_Sales"],
    errors="coerce"
)

flipkart["customer_rating"] = pd.to_numeric(
    flipkart["customer_rating"],
    errors="coerce")

# ===============================
# 4. Fill Quantity using Total_Sales / Unit_price
# ===============================

flipkart.loc[
    flipkart["Quantity"].isnull(),
    "Quantity"
] = (
    flipkart["Total_Sales"] /
    flipkart["Unit_price"]
)

flipkart["Quantity"] = flipkart["Quantity"].fillna(
    flipkart["Quantity"].median()
)

# Round quantity

flipkart["Quantity"] = flipkart["Quantity"].round()

# ===============================
# 3. Fill Total_Sales using Quantity * Unit_price
# ===============================

flipkart.loc[
    flipkart["Total_Sales"].isnull(),
    "Total_Sales"
] = (
    flipkart["Quantity"] *
    flipkart["Unit_price"]
)



# ===============================
# 5. Fill Product Category
# using Product Name mapping
# ===============================


category_map = (
    flipkart
    .dropna(subset=["Product_category"])
    .groupby("Product_Name")["Product_category"]
    .first()
)


flipkart["Product_category"] = flipkart.apply(
    lambda row:
    category_map[row["Product_Name"]]
    if pd.isna(row["Product_category"])
    and row["Product_Name"] in category_map
    else row["Product_category"],
    axis=1
)


# Remaining category missing

flipkart["Product_category"] = (
    flipkart["Product_category"]
    .fillna("Unknown")
)



# ===============================
# 6. Fill categorical columns
# ===============================


flipkart["Payment_method"] = (
    flipkart["Payment_method"]
    .fillna("Unknown")
)


flipkart["Country"] = (
    flipkart["Country"]
    .fillna("Unknown")
)



# ===============================
# 7. Fill customer rating
# ===============================


flipkart["customer_rating"] = (
    flipkart["customer_rating"]
    .fillna(
        flipkart["customer_rating"].median()
    )
)



# ===============================
# 8. Remove duplicates
# ===============================


flipkart = flipkart.drop_duplicates()



# ===============================
# 9. Final null check
# ===============================


print("Final Null Values")

print(flipkart.isnull().sum())



# ===============================
# 10. Check final shape
# ===============================

print("Final Shape:", flipkart.shape)
```

```python
Amazon.columns = Amazon.columns.str.strip()

Amazon = Amazon.rename(columns={
    "Order_ID": "order_id",
    "Product_Name": "Product_Name",
    "Product_Category": "Product_category",
    "Product_Name": "Product_Name",
    "Quantity": "Quantity",
    "Unit_Price_INR": "Unit_price",
    "Total_Sales_INR": "Total_Sales",
    "Payment_Method": "Payment_method",
    "Review_Rating": "customer_rating",
    "Country": "Country",
    "Date": "Order_Date",
    "Platform": "Platform",
    "City": "city",
    "Customer_ID":"customer_id",
})

print(Amazon.columns)
print(flipkart.columns)
```

```python
common_columns = Amazon.columns.intersection(flipkart.columns)

common_columns

Amazon_common = Amazon[common_columns]

flipkart_common = flipkart[common_columns]

final_data = pd.concat(
    [Amazon_common, flipkart_common],
    ignore_index=True
)
```

```python
# Convert Unknown to NaN and make numeric
final_data["customer_rating"] = pd.to_numeric(
    final_data["customer_rating"],
    errors="coerce"
)

# Find average rating
avg_rating = final_data["customer_rating"].mean()

print("Average Rating:", avg_rating)

# Fill missing ratings with average
final_data["customer_rating"] = final_data["customer_rating"].fillna(
    avg_rating
)

[col for col in final_data.columns if "date" in col.lower()]


final_data.isnull().sum()
```

```python
# Data Visualization

# Bar Chart - Total Sales Comparison (Amazon vs Flipkart)

platform_sales = final_data.groupby(
    "Platform"
)["Total_Sales"].sum()/ 10000000

plt.figure(figsize=(6,4))

platform_sales.plot(kind="bar")

plt.title("Total Sales Comparison")
plt.xlabel("Platform")
plt.ylabel("Total Sales (crores)")

plt.show()
```

```python
#  Pie Chart — Order Share by Platform

orders = final_data["Platform"].value_counts()

plt.figure(figsize=(6,6))
orders.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Order Contribution by Platform")
plt.ylabel("")

plt.show()
```

```python
# Quantity Vs Total Sales

plt.figure(figsize=(7,5))

plt.scatter(
    final_data["Quantity"],
    final_data["Total_Sales"]
)

plt.title("Quantity vs Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")

plt.show()
```

```python
#  Box Plot Rating distribution by Platform
plt.figure(figsize=(7,5))

sns.boxplot(
    x="Platform",
    y="customer_rating",
    data=final_data
)

plt.title("Customer Rating Distribution")

plt.show()
```

```python
# Horizontal Bar Chart — Top 10 Products by Sales

top_products = final_data.groupby(
    "Product_Name"
)["Total_Sales"].sum().sort_values(
    ascending=False
).head(10)/10000000


plt.figure(figsize=(8,5))

top_products.plot(
    kind="barh"
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales (crore)")

plt.show()
```

```python
#  Area Chart

# Remove extra spaces from column names
final_data.columns = final_data.columns.str.strip()


# Find date column
date_col = [col for col in final_data.columns if "date" in col.lower()]

print("Date column found:", date_col)


# Convert date column
final_data[date_col[0]] = pd.to_datetime(
    final_data[date_col[0]],
    dayfirst=True,
    errors="coerce"
)


# Create Month column
final_data["Month"] = final_data[date_col[0]].dt.month_name()


# Check result
final_data[[date_col[0], "Month"]].head()

# Month order
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

# Monthly sales
monthly_sales = final_data.groupby(
    "Month"
)["Total_Sales"].sum()

# Arrange months correctly
monthly_sales = monthly_sales.reindex(month_order)


# Convert to Crores
monthly_sales_crore = monthly_sales / 10000000


plt.figure(figsize=(10,5))

monthly_sales_crore.plot(
    kind="area"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (Crores INR)")

plt.xticks(rotation=45)

plt.show()
```

```python
#  Heatmap — Correlation Analysis

plt.figure(figsize=(7,5))

sns.heatmap(
    final_data[
        ["Quantity","Unit_price","Total_Sales","customer_rating"]
    ].corr(),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# 8.Stacked Bar Chart — Category Sales by Platform

category_platform = final_data.groupby(
    ["Product_category","Platform"]
)["Total_Sales"].sum().unstack()

category_platform_crore = category_platform / 10000000
category_platform.plot(
    kind="bar",
    stacked=True,
    figsize=(10,5)
)

plt.title("Category Sales Comparison")
plt.xlabel("Category")
plt.ylabel("Sales (crores)")

plt.xticks(rotation=45)

plt.show()
```
